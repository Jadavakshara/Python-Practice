```python
age = int(input())
health_score = int(input())
vehicle_type = input()
base = 10000

if age > 50:
    base *= 1.15
elif age < 25:
    base *= 1.20

if health_score >= 80:
    base *= 0.90
elif health_score < 60:
    base *= 1.20

if vehicle_type == "sports":
    base *= 1.30
elif vehicle_type == "suv":
    base *= 1.15

print(round(base,1))
```
