```python
units = int(input())
senior = input("Enter True or Not: ").lower()=="true"

if units <= 100:
    total = units * 1.5
elif units <= 200:
    total = units * 2.5
elif units <= 500:
    total = units * 4
elif units <= 800:
    total = units * 6
    
else:
    total = units * 6 * 1.05

if senior:
    total *= 0.9

print(total)
```
