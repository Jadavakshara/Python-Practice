```python
seat_type = input()
booking_days = int(input())
festival = input("Enter True or Not: ").lower()=="true"
age = int(input())
base_price = 5000

if (seat_type == "Business"):
    base_price *= 1.40
elif(seat_type == "Premium Economy"):
    base_price *= 1.20
    
if (booking_days>30):
    base_price *= 0.90
elif (booking_days<7):
    base_price *= 1.25

if festival:
    base_price *= 1.20

if age > 60:
    base_price *= 0.85

print("final ticket price = ",base_price)


```
